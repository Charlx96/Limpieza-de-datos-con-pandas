```python
import pandas as pd

df= pd.read_excel("/content/REGISTRO_RECLAMOS.xlsx")
df.head(3)
```

```python
df.info()
```

```python
#Lo que haremos es concatenar 2 columnas y mostrar
df["fecha de recepcion"]= df["FECHA RECEPCION"].astype(str) + " " + df["HORA RECEPCION"].astype(str)

df["fecha de respuesta"]= df["FECHA RESPUESTA"].astype(str) + " " + df["HORA RESPUESTA"].astype(str)
df.head(3)
```

```python
#borraremos las columnas que ya no se usaran

df.drop(["FECHA RECEPCION","HORA RECEPCION","FECHA RESPUESTA","HORA RESPUESTA"], axis=1, inplace=True)
df.head(3)
```

```python
df.info()
```

```python
#ahora debemos convertir el tipo de objetc a datatime
df["fecha de recepcion"] = pd.to_datetime(df["fecha de recepcion"], format="%d %m %Y %H %M %S")
df["fecha de respuesta"] = pd.to_datetime(df["fecha de respuesta"], format="%Y-%m-%d %H:%M:%S")

df.info()
```

```python
df["tiempo de atencion"] = df["fecha de respuesta"] - df["fecha de recepcion"]
df.head(3)
```

```python
df["respuesta en seg"] = df["tiempo de atencion"].dt.total_seconds()
df.head(3)
#aqui ya tenemos que tardo el cliente en ser atendido en dias tanto como en segundos
```

```python
df.describe()
#aqui puedo ver que la media es 135117,7 segundos,lo cual sigue siendo un valor alto de 37 horas para atender un reclamo
#tomaremos un reclamo rapido que sea menor a
```

```python
conteo = (df["respuesta en seg"] < 3600).value_counts()
print(conteo)
#Entonces los reclamos mas rapidos segun mi criterio fueron contestadas en menos de una hora (3600 segundos) un total de 336 personas
```
